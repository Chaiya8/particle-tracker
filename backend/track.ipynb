```python
from tracker import Tracker
import matplotlib.pyplot as plt
```

```python
tracker = Tracker(root_folder="D:\\OneDrive\\Desktop\\All\\magnetlab\\20230718\\",
                  detection_model="models\\experimental\\weights",
                  scale=0.294,
                  fps = 56.95)
```

```python
cutoff = 0.999
```

```python
sample = tracker.random_frame()
sample_detections = tracker.detect_frame(sample, cutoff=cutoff)
plt.axis("off")
plt.imshow(sample, cmap="gray")
plt.scatter(sample_detections[:, 1], sample_detections[:, 0], color="r", s=0.5)
# plt.savefig("sample.png", bbox_inches="tight", pad_inches=0)
```

```python
tracker.detect_all(cutoff=cutoff)
```

```python
tracker.track_all(min_duration=20, min_displacement=50, search_range=25, memory=0)
```

```python
f_to_v = tracker.freq_to_v()
v_vs_f = tracker.compute_stats(f_to_v)
```

```python
import matplotlib.pyplot as plt
fig = plt.figure()
ax = fig.add_subplot(111)
tracker.plot_v_vs_f(ax, v_vs_f)
```
